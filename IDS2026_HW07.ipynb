**c. First shots at visualizing the data**

Take a look at the data as plots to get more insight to the patterns, trends, and relationships. This could be histograms, scatter plots, box plots, ect. What did you gather? What stands out?

```python
## TODO
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(13, 9))

# Whats the dropout rate?
counts = data["dropout"].value_counts().reindex([0, 1])
axes[0, 0].bar(["No dropout (0)", "Dropout (1)"], counts, color=["steelblue", "orange"])
axes[0, 0].set_title("Dropout distribution")
axes[0, 0].set_ylabel("Number of students")
for i, count in enumerate(counts):
    axes[0, 0].text(i, count + 500, f"{count:,} ({count / len(data):.1%})", ha="center")

# Compare GPA distributions of students who stayed vs dropped out
for status, label, color in [(0, "No dropout", "steelblue"), (1, "Dropout", "orange")]:
    axes[0, 1].hist(data.loc[data["dropout"] == status, "gpa_semester_1"], bins=25, alpha=0.6, density=True, label=label, color=color)
axes[0, 1].set_title("First semester GPA by dropout status")
axes[0, 1].set_xlabel("GPA (0-4)")
axes[0, 1].set_ylabel("Density")
axes[0, 1].legend()

# Compare dropout rate by attendance
attendance_groups = pd.cut(data["attendance_rate"], bins=[-0.01, 40, 60, 80, 100], labels=["0-40%", "40-60%", "60-80%", "80-100%"])
attendance_dropout = (data.groupby(attendance_groups, observed=True)["dropout"].mean() * 100)

axes[1, 0].bar(attendance_dropout.index.astype(str), attendance_dropout.values, color="steelblue")
axes[1, 0].set_title("Dropout rate by attendance")
axes[1, 0].set_xlabel("Attendance range")
axes[1, 0].set_ylabel("Dropout rate (%)")
for i, rate in enumerate(attendance_dropout):
    axes[1, 0].text(i, rate+0.5, f"{rate:.1f}%", ha="center")

# Compare dropout rates based on faimily income groups.
income_order = ["Low", "Middle", "High"]
income_dropout = (data.groupby("family_income_level")["dropout"].mean() * 100).reindex(income_order)

axes[1, 1].bar(income_dropout.index, income_dropout.values, color="steelblue")
axes[1, 1].set_title("Dropout rate by family income")
axes[1, 1].set_xlabel("Family income level")
axes[1, 1].set_ylabel("Dropout rate (%)")
for i, rate in enumerate(income_dropout):
    axes[1, 1].text(i, rate, f"{rate:.1f}%", ha="center")

plt.tight_layout()
plt.show()
```

**Answer:** About 32% of students dropped out. Students who dropped out generally have a lower first semester GPA, other big factors in dropout rate seems to be attendance rate and family income level.

**d. Check for data quality**

Missing values, duplicates, impossible or inconsistent values. Also watch out for outliers and anomalies that may require further investigation. Explain what you found and what you are going to do with the information.


```python

missing_values = data.isnull().sum().copy()
missing_values = missing_values[missing_values > 0]
print(f"There are {missing_values.shape[0]} features with missing values\n")

# Outliers, based on IQR
numeric = data.select_dtypes(include="number").copy()

counter = 0
for label, col in numeric.items():
    # Skip binary columns
    if col.isin([0, 1]).all():
        continue

    stats = col.describe()

    Q1 = stats["25%"]
    Q3 = stats["75%"]
    
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = col[(col < lower_bound) | (col > upper_bound)]
    amount = outliers.shape[0]

    if(amount > 0):
        print(f"There are {amount:>5} outliers for feature '{label}'")
        counter += 1

print(f"\nTotal features with potential outliers: {counter}\n")


non_numeric = data.select_dtypes(exclude="number").copy()

for label, col in non_numeric.items():
    print(f"Value counts for {label}: \n{col.value_counts()}\n")
```

**Answer:**
According to Kaggle, there are no missing values in the dataset, and we confirmed this to be true. <br>
IQR was calculated on numeric, non-boolean values. There are 13 features with potential outliers, which definitely need to be checked. <br>
All non-numeric values were counted, and there are some observations:<br>
-- All of them turned out to be categorical; there are no "hidden" numeric features.<br>
-- There are no typos.<br>
-- The distribution of every categoric feature is imbalanced. This could need some normalization if these features turn out to be significant.<br>
